```python
import pandas as pd
from hogsboutils import read_contacts

contacts = read_contacts()
common_room_cost = 23706
```

```python
df_rent = pd.read_csv('rent.csv', sep=';')
```

```python
df_rent['square_m'] = df_rent['Area (m2)'].str.split(' ', expand=True)[0].astype(int)
```

```python
contacts['Lantmäterinummer'] = contacts['Familjebostäder\nLgh/Objektnummer\n2120-00xx'].str.split('/', expand=True)[0].astype(int)
df_res_count = contacts.groupby(['Lantmäterinummer']).count()[['Namn / Name']].rename({'Namn / Name': 'num_residents'}, axis=1)
df = pd.merge_asof(df_rent, df_res_count, left_on='Lantmäterinummer', right_index=True)
```

```python
common_room_cost_month = 23706
df['cost_flat'] = common_room_cost / len(df) / df['num_residents']
df['cost_per_resident'] = common_room_cost_month * df['num_residents'] / df['num_residents'].sum() /  df['num_residents']
df['cost_square_m'] = common_room_cost_month * df['square_m'] / df['square_m'].sum() /  df['num_residents']
df['cost_rooms'] = common_room_cost_month * df['Rooms'] / df['Rooms'].sum() /  df['num_residents']

for cost_col in ['cost_flat', 'cost_per_resident', 'cost_square_m','cost_rooms']:
    df[cost_col] = df[cost_col].astype(int)
```
